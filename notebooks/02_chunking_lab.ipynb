# ✦ LumenTrail: inspect chunk boundaries

This exercise uses synthetic text, not health guidance. A fixed width is a useful baseline, but it can cut a statement from its warning. Paragraph boundaries are easier to inspect. Neither choice guarantees a good retrieval result.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'src'))
from lumentrail.chunking import fixed_character_chunks, paragraph_chunks

source_text = (
    'AF is an irregular heart rhythm. This is a synthetic teaching sentence.\n\n'
    'A caution belongs with the statement it qualifies. Keep both in the same passage.'
)
fixed = fixed_character_chunks(source_text, 80)
structured = paragraph_chunks(source_text, 100)
print('Fixed width:', fixed)
print('Whole paragraphs:', structured)
```

The paragraph method refuses a paragraph longer than the chosen limit. That asks you to review the source rather than silently cut a caution in half. A semantic chunker can be another candidate, but it needs an embedding model and may still split the wrong fact. Compare chunk choices using reviewed questions, source versions and the same retrieval settings.

```python
assert len(structured) == 2
assert any('A caution belongs' in chunk and 'same passage' in chunk for chunk in structured)
assert ''.join(fixed) == source_text
print('The source is intact; review which boundaries help the search task.')
```
